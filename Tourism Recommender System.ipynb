```python
# Drop Useless Columns
print("\n" + "="*60)
print("DROPPING USELESS COLUMNS")
print("="*60)

# List of Columns to Drop
columns_to_drop = []

# Columns With >80% missing
for col in df.columns:
    if df[col].isnull().sum() / len(df) > 0.8:
        columns_to_drop.append(col)

# Constant Columns
for col in df.columns:
    if df[col].dtype in ['int64', 'float64', 'object']:
        if df[col].value_counts(normalize=True).max() > 0.95:
            columns_to_drop.append(col)

# Columns Specifically for Group Tours (Too Sparse)
group_tour_cols = ['group_fee', 'trip_days', 'main_spots', 'transport_mode']
for col in group_tour_cols:
    if col in df.columns:
        columns_to_drop.append(col)

# Remove Duplicates from Drop List
columns_to_drop = list(set(columns_to_drop))

print("\nColumns to be dropped ({len(columns_to_drop)}):")
for col in sorted(columns_to_drop):
    if col in df.columns:
        print(f"   - {col} (missing: {df[col].isnull().sum()/len(df)*100:.1f}%)")
    else:
        print(f"   - {col} (already dropped)")

# Drop columns
df_cleaned = df.drop(columns=columns_to_drop, errors='ignore')

print(f"\nBefore dropping: {df.shape[1]} columns")
print(f"After dropping: {df_cleaned.shape[1]} columns")
print(f"Dropped: {df.shape[1] - df_cleaned.shape[1]} columns")

# Show Remaining Columns
print("\n Remaining columns:")
for col in df_cleaned.columns:
    print(f"   - {col}")
```

```python
# Setting for Modeling
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from surprise import SVD, Dataset, Reader
from sklearn.decomposition import TruncatedSVD
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    accuracy_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

RANDOM_SEED = 42
TEST_FRACTION = 0.20

SVD_FACTORS = 20
SVD_EPOCHS = 25

TOP_K_VALUES = [5, 10]

# Rating >= 4 Means the User Liked the Attraction
LIKE_THRESHOLD = 4.0

# Hybrid Weights
ALPHA_VALUES = np.arange(0.0, 1.01, 0.10)
```

```python
# Data Preparation for Modeling
print("=" * 60)
print("Preparing Data")
print("=" * 60)

ratings = df[
    ["tourist_id", "attraction_name", "rating"]
].copy()

# Average Duplicate Ratings per Tourist-Item Pair
ratings = (
    ratings
    .groupby(
        ["tourist_id", "attraction_name"],
        as_index=False
    )["rating"]
    .mean()
)

# Canonical Mappings
user_ids = sorted(
    ratings["tourist_id"].unique()
)

item_ids = sorted(
    ratings["attraction_name"].unique()
)

user_to_idx = {
    user_id: i
    for i, user_id in enumerate(user_ids)
}

item_to_idx = {
    item_id: i
    for i, item_id in enumerate(item_ids)
}

idx_to_item = {
    i: item
    for item, i in item_to_idx.items()
}

ratings["user_idx"] = ratings[
    "tourist_id"
].map(user_to_idx)

ratings["item_idx"] = ratings[
    "attraction_name"
].map(item_to_idx)

n_users = len(user_ids)
n_items = len(item_ids)

density = len(ratings) / (
    n_users * n_items
)

print(f"Users:            {n_users:,}")
print(f"Items:            {n_items:,}")
print(f"Interactions:     {len(ratings):,}")
print(f"Matrix density:   {density:.4%}")
print(f"Matrix sparsity:  {1 - density:.4%}")
```

```python
# Train and Test Data Splitting
print("=" * 60)
print("Train and Test Data Splitting")
print("=" * 60)

train_rows = []
test_rows = []

for uid, group in ratings.groupby("user_idx"):
    n = len(group)
    # Users with Only One Interaction
    # Remain Entirely in Training
    if n < 2:
        train_rows.append(group)
        continue
    # At Least One Test Item
    n_test = max(
        1,
        int(round(n * TEST_FRACTION))
    )
    shuffled = group.sample(
        frac=1.0,
        random_state=RANDOM_SEED
    )
    test_rows.append(
        shuffled.iloc[:n_test]
    )
    train_rows.append(
        shuffled.iloc[n_test:]
    )

train_df = pd.concat(
    train_rows,
    ignore_index=True
)

test_df = (
    pd.concat(
        test_rows,
        ignore_index=True
    )
    if test_rows
    else pd.DataFrame(
        columns=ratings.columns
    )
)

global_mean = train_df["rating"].mean()
print(f"Train: {len(train_df):,}")
print(f"Test:  {len(test_df):,}")
print(f"Global mean rating: {global_mean:.4f}")
```

```python
# Model 1:
# SVD COLLABORATIVE FILTERING

print("\n" + "=" * 60)
print("Model 1: SVD COLLABORATIVE FILTERING")
print("=" * 60)

# 3.1 Training SVD Model
print("\nTraining SVD model...")

reader = Reader(rating_scale=(1.0, 5.0))

data = Dataset.load_from_df(
    train_df[["user_idx", "item_idx", "rating"]], reader
)

trainset = data.build_full_trainset()

svd_model = SVD(
    n_factors=SVD_FACTORS, random_state=RANDOM_SEED, n_epochs=SVD_EPOCHS
)

svd_model.fit(trainset)

print(f"CF component ready (Surprise SVD k={SVD_FACTORS})")

# Predict Test Rating
print("\nGenerating SVD predictions for test data...")

svd_test_predictions = []

for row in test_df.itertuples():
    prediction = svd_model.predict(row.user_idx, row.item_idx)
    svd_test_predictions.append(prediction.est)

svd_test_predictions = np.clip(np.array(svd_test_predictions), 1.0, 5.0)

y_true = test_df["rating"].values
y_pred = svd_test_predictions

# Rating Prediction Metrics
rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mae = mean_absolute_error(y_true, y_pred)

# Classification:
# Rating >= 4.0 = Like
# Rating < 4.0 = Not Like

LIKE_THRESHOLD = 4.0

y_true_class = (y_true >= LIKE_THRESHOLD).astype(int)
y_pred_class = (y_pred >= LIKE_THRESHOLD).astype(int)

class_accuracy = accuracy_score(y_true_class, y_pred_class)
class_f1 = f1_score(y_true_class, y_pred_class, zero_division=0)

print("\n" + "-" * 60)
print("SVD Rating Prediction Result")
print("-" * 60)

print(f"RMSE           = {rmse:.4f}")
print(f"MAE            = {mae:.4f}")
print(f"Class Accuracy = {class_accuracy:.4f}")
print(f"Class F1-Score = {class_f1:.4f}")
print("\n")

# Confusion Matrix
cm = confusion_matrix(y_true_class, y_pred_class)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm, display_labels=["Not Like", "Like"]
)

fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(ax=ax, values_format="d", colorbar=True, cmap=plt.cm.Blues)

ax.grid(False)
plt.title("MLP: Confusion Matrix", fontsize=12, pad=10)
plt.tight_layout()
plt.show()

print("\n")

# Actual vs Predicted Ratings
plt.figure(figsize=(7, 6))

plt.scatter(y_true, y_pred, alpha=0.25)
plt.plot([1, 5], [1, 5], linestyle="--")

plt.xlabel("Actual Rating")
plt.ylabel("Predicted Rating")
plt.title("SVD: Actual vs Predicted Ratings")

plt.xlim(1, 5)
plt.ylim(1, 5)

plt.tight_layout()
plt.show()

# Create Full User-Item Prediction Matrix
print("\nGenerating full SVD prediction matrix...")

pred_cf_matrix = np.zeros(
    (n_users, n_items)
)

for u in range(n_users):

    for i in range(n_items):

        pred_cf_matrix[u, i] = (
            svd_model
            .predict(u, i)
            .est
        )


pred_cf_matrix = np.clip(
    pred_cf_matrix,
    1.0,
    5.0
)

print(
    "SVD prediction matrix shape:",
    pred_cf_matrix.shape
)


# Top-N Evaluation For SVD Only
TOP_K = 5

print("\n" + "-" * 60)
print(f"SVD TOP-{TOP_K} RECOMMENDATION EVALUATION")
print("-" * 60)

# Items already seen in training
train_items_by_user = (
    train_df.groupby("user_idx")["item_idx"].apply(set).to_dict()
)

# Relevant test items
# Rating >= 4.0
test_relevant_by_user = (
    test_df[test_df["rating"] >= LIKE_THRESHOLD]
    .groupby("user_idx")["item_idx"]
    .apply(set)
    .to_dict()
)

# Evaluation containers
precision_scores = []
recall_scores = []
f1_scores = []
hit_scores = []
ndcg_scores = []
evaluated_users = 0

# Evaluate every user
for user_idx, relevant_items in test_relevant_by_user.items():
    if user_idx not in train_items_by_user:
        continue

    seen_items = train_items_by_user[user_idx]

    # Candidate items = items NOT seen during training
    candidate_items = [i for i in range(n_items) if i not in seen_items]

    if len(candidate_items) == 0:
        continue

    # Rank candidate items using SVD predicted rating
    candidate_scores = [
        (i, pred_cf_matrix[user_idx, i]) for i in candidate_items
    ]

    candidate_scores.sort(key=lambda x: x[1], reverse=True)

    recommended_items = [item for item, score in candidate_scores[:TOP_K]]
    recommended_set = set(recommended_items)
    relevant_set = set(relevant_items)

    # Number of hits
    hits = len(recommended_set.intersection(relevant_set))

    # Precision@K
    precision = hits / TOP_K

    # Recall@K
    recall = hits / len(relevant_set) if len(relevant_set) > 0 else 0

    # F1@K
    if precision + recall > 0:
        f1 = 2 * precision * recall / (precision + recall)
    else:
        f1 = 0

    # Hit Rate@K
    hit = 1 if hits > 0 else 0

    # NDCG@K
    dcg = 0.0

    for rank, item in enumerate(recommended_items, start=1):
        if item in relevant_set:
            dcg += 1 / np.log2(rank + 1)

    ideal_hits = min(len(relevant_set), TOP_K)

    idcg = sum(1 / np.log2(rank + 1) for rank in range(1, ideal_hits + 1))

    ndcg = dcg / idcg if idcg > 0 else 0

    # Store results
    precision_scores.append(precision)
    recall_scores.append(recall)
    f1_scores.append(f1)
    hit_scores.append(hit)
    ndcg_scores.append(ndcg)

    evaluated_users += 1

# Display Top-N Results
svd_precision = np.mean(precision_scores)
svd_recall = np.mean(recall_scores)
svd_f1 = np.mean(f1_scores)
svd_hr = np.mean(hit_scores)
svd_ndcg = np.mean(ndcg_scores)

print(f"Users evaluated = {evaluated_users}")
print(f"Precision@{TOP_K} = {svd_precision:.4f}")
print(f"Recall@{TOP_K}    = {svd_recall:.4f}")
print(f"F1@{TOP_K}        = {svd_f1:.4f}")
print(f"HR@{TOP_K}        = {svd_hr:.4f}")
print(f"NDCG@{TOP_K}      = {svd_ndcg:.4f}")

# Sample Tourist Recommendation
print("\n" + "=" * 60)
print("Sample Tourist Recommendation")
print("=" * 60)

# Select a representative tourist
user_interaction_counts = train_df.groupby("user_idx").size()

# Select a Tourist with 8-12 Previous Interactions.
eligible_users = user_interaction_counts[
    (user_interaction_counts >= 8) & (user_interaction_counts <= 12)
]

# Randomly Select one Tourist
SAMPLE_USER_IDX = eligible_users.sample(
    1, random_state=RANDOM_SEED
).index[0]

# Convert user_idx Back to original tourist_id
tourist_id = user_ids[SAMPLE_USER_IDX]

print(f"\nTourist ID: {tourist_id}")
print(
    f"Previous interactions: "
    f"{user_interaction_counts[SAMPLE_USER_IDX]}"
)

# Find previous attractions
previous_items = train_df[
    train_df["user_idx"] == SAMPLE_USER_IDX
].sort_values("rating", ascending=False)

print("\nPrevious Attractions")
print("-" * 70)
print(f"{'Attraction':<45}{'Rating':>10}")
print("-" * 70)

for row in previous_items.itertuples():
    attraction_name = idx_to_item[row.item_idx]
    print(f"{attraction_name:<45}{row.rating:>10.2f}")

# Get attractions already visited
seen_items = set(previous_items["item_idx"])

# Generate candidate attractions
candidate_items = [
    item_idx for item_idx in range(n_items) if item_idx not in seen_items
]

# Predict ratings for candidate attractions
recommendation_scores = []

for item_idx in candidate_items:
    predicted_rating = pred_cf_matrix[SAMPLE_USER_IDX, item_idx]
    recommendation_scores.append((item_idx, predicted_rating))

# Sort attractions by predicted rating
recommendation_scores.sort(key=lambda x: x[1], reverse=True)

# Select Top 5 recommendations
top_recommendations = recommendation_scores[:5]

# Display recommendations
print("\nRecommended Attractions")
print("-" * 70)

print(
    f"{'Rank':<8}"
    f"{'Attraction':<45}"
    f"{'Predicted Rating':>17}"
)
print("-" * 70)

for rank, (item_idx, predicted_rating) in enumerate(
    top_recommendations, start=1
):
    attraction_name = idx_to_item[item_idx]

    print(
        f"{rank:<8}"
        f"{attraction_name:<45}"
        f"{predicted_rating:>17.2f}"
    )

# Create a recommendation DataFrame
recommendation_results = []

for rank, (item_idx, predicted_rating) in enumerate(
    top_recommendations, start=1
):
    recommendation_results.append({
        "Tourist ID": tourist_id,
        "Rank": rank,
        "Attraction": idx_to_item[item_idx],
        "Predicted Rating": round(predicted_rating, 2),
    })

recommendation_results_df = pd.DataFrame(recommendation_results)

# Final SVD Model Summary
print("\n" + "=" * 60)
print("SVD FINAL SUMMARY")
print("=" * 60)

print(f"RMSE            : {rmse:.4f}")
print(f"MAE             : {mae:.4f}")
print(f"Classification Accuracy : {class_accuracy:.4f}")
print(f"Classification F1       : {class_f1:.4f}")
print(f"Precision@5     : {svd_precision:.4f}")
print(f"Recall@5        : {svd_recall:.4f}")
print(f"F1@5            : {svd_f1:.4f}")
print(f"HR@5            : {svd_hr:.4f}")
print(f"NDCG@5          : {svd_ndcg:.4f}")
print("=" * 60)
```

```python
# Model 2:
# TF-IDF CONTENT-BASED RECOMMENDATION

print("\n" + "=" * 60)
print("Model 2: TF-IDF CONTENT-BASED RECOMMENDATION")
print("=" * 60)

# Building Attraction Metadata & TF-IDF Features
print("\nBuilding attraction metadata and TF-IDF features...")

def price_tier(price):
    if pd.isna(price):
        return "Unknown"
    elif price <= 0:
        return "Free"
    elif price <= 60:
        return "Budget"
    elif price <= 150:
        return "MidRange"
    else:
        return "Premium"

attr_meta = (
    df.groupby("attraction_name")
    .agg(
        attraction_category=(
            "attraction_category",
            lambda x: x.mode().iloc[0],
        ),
        attraction_level=(
            "attraction_level",
            lambda x: x.mode().iloc[0],
        ),
        city=("city", lambda x: x.mode().iloc[0]),
        province=("province", lambda x: x.mode().iloc[0]),
        ticket_price=("ticket_price", "mean"),
    )
    .reset_index()
)

attr_meta["price_tier"] = attr_meta["ticket_price"].apply(price_tier)
attr_meta["item_idx"] = attr_meta["attraction_name"].map(item_to_idx)

attr_meta = attr_meta.sort_values("item_idx").reset_index(drop=True)

text_columns = [
    "attraction_category",
    "attraction_level",
    "province",
    "city",
    "price_tier",
]

for col in text_columns:
    attr_meta[col] = (
        attr_meta[col]
        .fillna("Unknown")
        .astype(str)
        .str.replace(" ", "_", regex=False)
    )

# Weighted Metadata Soup (Category = 3x, City = 2x, Others = 1x)
attr_meta["soup"] = (
    (attr_meta["attraction_category"] + " ") * 3
    + (attr_meta["city"] + " ") * 2
    + attr_meta["attraction_level"] + " "
    + attr_meta["province"] + " "
    + attr_meta["price_tier"]
)

vectorizer = TfidfVectorizer(token_pattern=r"[^\s]+")
tfidf_matrix = vectorizer.fit_transform(attr_meta["soup"]).toarray()

print(f"TF-IDF matrix shape: {tfidf_matrix.shape}")

# Building User Content Profiles & Predictions
print("\nBuilding user content profiles and cosine similarity matrix...")

user_profile_matrix = np.zeros((n_users, tfidf_matrix.shape[1]))

for uid, group in train_df.groupby("user_idx"):
    item_indices = group["item_idx"].values
    ratings_weights = group["rating"].values

    user_profile_matrix[uid] = np.average(
        tfidf_matrix[item_indices], axis=0, weights=ratings_weights
    )

content_sim_matrix = cosine_similarity(user_profile_matrix, tfidf_matrix)

# User mean ratings for scaling continuous predictions
user_sums = np.zeros(n_users)
user_counts = np.zeros(n_users)

for uid, group in train_df.groupby("user_idx"):
    user_sums[uid] = group["rating"].sum()
    user_counts[uid] = len(group)

user_means = np.divide(
    user_sums,
    user_counts,
    out=np.full(n_users, global_mean),
    where=user_counts > 0,
)

pred_content_matrix = user_means[:, None] + (
    content_sim_matrix - content_sim_matrix.mean(axis=1, keepdims=True)
) * 1.5

pred_content_matrix = np.clip(pred_content_matrix, 1.0, 5.0)

print("Content-based component ready")

# Predict Test Rating
print("\nGenerating TF-IDF content predictions for test data...")

content_test_predictions = []

for row in test_df.itertuples():
    pred_val = pred_content_matrix[row.user_idx, row.item_idx]
    content_test_predictions.append(pred_val)

content_test_predictions = np.clip(
    np.array(content_test_predictions), 1.0, 5.0
)

y_true = test_df["rating"].values
y_pred = content_test_predictions

# Rating Prediction Metrics
rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mae = mean_absolute_error(y_true, y_pred)

# Classification (Rating >= 4.0 = Like, Rating < 4.0 = Not Like)
LIKE_THRESHOLD = 4.0

y_true_class = (y_true >= LIKE_THRESHOLD).astype(int)
y_pred_class = (y_pred >= LIKE_THRESHOLD).astype(int)

class_accuracy = accuracy_score(y_true_class, y_pred_class)
class_f1 = f1_score(y_true_class, y_pred_class, zero_division=0)

print("\n" + "-" * 60)
print("TF-IDF Content Rating Prediction Result")
print("-" * 60)
print(f"RMSE           = {rmse:.4f}")
print(f"MAE            = {mae:.4f}")
print(f"Class Accuracy = {class_accuracy:.4f}")
print(f"Class F1-Score = {class_f1:.4f}")
print("\n")

# Confusion Matrix
cm = confusion_matrix(y_true_class, y_pred_class)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm, display_labels=["Not Like", "Like"]
)

fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(
    ax=ax, values_format="d", colorbar=True, cmap=plt.cm.Blues
)

ax.grid(False)
plt.title("TF-IDF: Confusion Matrix", fontsize=12, pad=10)
plt.tight_layout()
plt.show()

print("\n")

# Actual vs Predicted Ratings Scatter Plot
plt.figure(figsize=(7, 6))
plt.scatter(y_true, y_pred, alpha=0.25, color="#1f77b4")
plt.plot([1, 5], [1, 5], linestyle="--", color="red")
plt.xlabel("Actual Rating")
plt.ylabel("Predicted Rating")
plt.title("TF-IDF Content: Actual vs Predicted Ratings")
plt.xlim(1, 5)
plt.ylim(1, 5)
plt.tight_layout()
plt.show()

# Top-N Evaluation For TF-IDF Content Only
TOP_K = 5

print("\n" + "-" * 60)
print(f"TF-IDF CONTENT TOP-{TOP_K} RECOMMENDATION EVALUATION")
print("-" * 60)

# Items already seen in training
train_items_by_user = (
    train_df.groupby("user_idx")["item_idx"].apply(set).to_dict()
)

# Relevant test items (Rating >= 4.0)
test_relevant_by_user = (
    test_df[test_df["rating"] >= LIKE_THRESHOLD]
    .groupby("user_idx")["item_idx"]
    .apply(set)
    .to_dict()
)

# Evaluation containers
precision_scores = []
recall_scores = []
f1_scores = []
hit_scores = []
ndcg_scores = []
evaluated_users = 0

# Evaluate every user
for user_idx, relevant_items in test_relevant_by_user.items():

    if user_idx not in train_items_by_user:
        continue

    seen_items = train_items_by_user[user_idx]

    # Candidate items = items NOT seen during training
    candidate_items = [i for i in range(n_items) if i not in seen_items]

    if len(candidate_items) == 0:
        continue

    # Rank candidate items using TF-IDF predicted rating
    candidate_scores = [
        (i, pred_content_matrix[user_idx, i]) for i in candidate_items
    ]

    candidate_scores.sort(key=lambda x: x[1], reverse=True)

    recommended_items = [
        item for item, score in candidate_scores[:TOP_K]
    ]

    recommended_set = set(recommended_items)
    relevant_set = set(relevant_items)

    # Number of hits
    hits = len(recommended_set.intersection(relevant_set))

    # Precision@K
    precision = hits / TOP_K

    # Recall@K
    recall = hits / len(relevant_set) if len(relevant_set) > 0 else 0

    # F1@K
    if precision + recall > 0:
        f1 = 2 * precision * recall / (precision + recall)
    else:
        f1 = 0

    # Hit Rate@K
    hit = 1 if hits > 0 else 0

    # NDCG@K
    dcg = 0.0
    for rank, item in enumerate(recommended_items, start=1):
        if item in relevant_set:
            dcg += 1 / np.log2(rank + 1)

    ideal_hits = min(len(relevant_set), TOP_K)
    idcg = sum(1 / np.log2(rank + 1) for rank in range(1, ideal_hits + 1))

    ndcg = dcg / idcg if idcg > 0 else 0

    # Store results
    precision_scores.append(precision)
    recall_scores.append(recall)
    f1_scores.append(f1)
    hit_scores.append(hit)
    ndcg_scores.append(ndcg)

    evaluated_users += 1

# Display Top-N Results
content_precision = np.mean(precision_scores)
content_recall = np.mean(recall_scores)
content_f1 = np.mean(f1_scores)
content_hr = np.mean(hit_scores)
content_ndcg = np.mean(ndcg_scores)

print(f"Users evaluated = {evaluated_users}")
print(f"Precision@{TOP_K} = {content_precision:.4f}")
print(f"Recall@{TOP_K}    = {content_recall:.4f}")
print(f"F1@{TOP_K}        = {content_f1:.4f}")
print(f"HR@{TOP_K}        = {content_hr:.4f}")
print(f"NDCG@{TOP_K}      = {content_ndcg:.4f}")

# Sample Tourist Recommendation
print("\n" + "=" * 60)
print("Sample Tourist Recommendation")
print("=" * 60)

# Select a representative tourist
user_interaction_counts = train_df.groupby("user_idx").size()

# Select a Tourist with 8-12 Previous Interactions.
eligible_users = user_interaction_counts[
    (user_interaction_counts >= 8) & (user_interaction_counts <= 12)
]

# Randomly Select one Tourist
SAMPLE_USER_IDX = eligible_users.sample(
    1, random_state=RANDOM_SEED
).index[0]

# Convert user_idx Back to original tourist_id
tourist_id = user_ids[SAMPLE_USER_IDX]

print(f"\nTourist ID: {tourist_id}")
print(
    f"Previous interactions: {user_interaction_counts[SAMPLE_USER_IDX]}"
)

# Find Previous Attractions
previous_items = train_df[
    train_df["user_idx"] == SAMPLE_USER_IDX
].sort_values("rating", ascending=False)

print("\nPrevious Attractions")
print("-" * 70)
print(f"{'Attraction':<45}{'Rating':>10}")
print("-" * 70)

for row in previous_items.itertuples():
    attraction_name = idx_to_item[row.item_idx]
    print(f"{attraction_name:<45}{row.rating:>10.2f}")

# Get Attractions Already Visited
seen_items = set(previous_items["item_idx"])

# Generate Candidate Attractions
candidate_items = [
    item_idx for item_idx in range(n_items) if item_idx not in seen_items
]

# Predict Ratings for Candidate Attractions
recommendation_scores = []

for item_idx in candidate_items:
    predicted_rating = pred_content_matrix[SAMPLE_USER_IDX, item_idx]
    recommendation_scores.append((item_idx, predicted_rating))

# Sort Attractions by Predicted Rating
recommendation_scores.sort(key=lambda x: x[1], reverse=True)

# Select Top 5 Recommendations
top_recommendations = recommendation_scores[:5]

# Display Recommendations
print("\nRecommended Attractions")
print("-" * 70)
print(
    f"{'Rank':<8}{'Attraction':<45}{'Predicted Rating':>17}"
)
print("-" * 70)

for rank, (item_idx, predicted_rating) in enumerate(
    top_recommendations, start=1
):
    attraction_name = idx_to_item[item_idx]
    print(
        f"{rank:<8}{attraction_name:<45}{predicted_rating:>17.2f}"
    )

# Create a Recommendation DataFrame
recommendation_results = []

for rank, (item_idx, predicted_rating) in enumerate(
    top_recommendations, start=1
):
    recommendation_results.append(
        {
            "Tourist ID": tourist_id,
            "Rank": rank,
            "Attraction": idx_to_item[item_idx],
            "Predicted Rating": round(predicted_rating, 2),
        }
    )
recommendation_results_df = pd.DataFrame(recommendation_results)

# Final TF-IDF Model Summary
print("\n" + "=" * 60)
print("TF-IDF FINAL SUMMARY")
print("=" * 60)
print(f"RMSE                    : {rmse:.4f}")
print(f"MAE                     : {mae:.4f}")
print(f"Classification Accuracy : {class_accuracy:.4f}")
print(f"Classification F1       : {class_f1:.4f}")
print(f"Precision@5             : {content_precision:.4f}")
print(f"Recall@5                : {content_recall:.4f}")
print(f"F1@5                    : {content_f1:.4f}")
print(f"HR@5                    : {content_hr:.4f}")
print(f"NDCG@5                  : {content_ndcg:.4f}")
print("=" * 60)
```

```python
# Model 3:
# NEURAL NETWORK (MLP + EMBEDDINGS)

print("\n" + "=" * 60)
print("Model 3: NEURAL NETWORK (MLP + EMBEDDINGS)")
print("=" * 60)

# Training MLP Model & Feature Preparation
print("\nTraining MLP model on latent factors and biases...")

# Item mean ratings as bias feature
item_sums_nn = np.zeros(n_items)
item_counts_nn = np.zeros(n_items)

for iid, group in train_df.groupby("item_idx"):
    item_sums_nn[iid] = group["rating"].sum()
    item_counts_nn[iid] = len(group)

item_means_nn = np.divide(
    item_sums_nn,
    item_counts_nn,
    out=np.full(n_items, global_mean),
    where=item_counts_nn > 0,
)

# Extract CF latent factors directly from trained Surprise SVD model
user_cf_embed = np.zeros((n_users, SVD_FACTORS))
item_cf_embed = np.zeros((n_items, SVD_FACTORS))

for uid in range(n_users):
    try:
        inner_uid = trainset.to_inner_uid(uid)
        user_cf_embed[uid] = svd_model.pu[inner_uid]
    except ValueError:
        pass  # leave as zeros for cold-start user

for iid in range(n_items):
    try:
        inner_iid = trainset.to_inner_iid(iid)
        item_cf_embed[iid] = svd_model.qi[inner_iid]
    except ValueError:
        pass  # leave as zeros for cold-start item

# Reduce TF-IDF item vectors to a small dense content embedding
CONTENT_EMBED_DIM = 10

content_svd = TruncatedSVD(
    n_components=CONTENT_EMBED_DIM, random_state=RANDOM_SEED
)

item_content_embed = content_svd.fit_transform(tfidf_matrix)


# Build (user, item) -> feature vector function
def build_nn_features(user_idx_arr, item_idx_arr):
    uf = user_cf_embed[user_idx_arr]
    itf = item_cf_embed[item_idx_arr]
    icf = item_content_embed[item_idx_arr]

    ub = (user_means[user_idx_arr] - global_mean).reshape(-1, 1)
    ib = (item_means_nn[item_idx_arr] - global_mean).reshape(-1, 1)

    return np.hstack([uf, itf, icf, ub, ib])


X_train_nn = build_nn_features(
    train_df["user_idx"].values, train_df["item_idx"].values
)

y_train_nn = train_df["rating"].values

nn_scaler = StandardScaler()
X_train_nn_scaled = nn_scaler.fit_transform(X_train_nn)

mlp_model = MLPRegressor(
    hidden_layer_sizes=(64, 32),
    activation="relu",
    solver="adam",
    alpha=1e-3,
    learning_rate_init=1e-3,
    max_iter=300,
    early_stopping=True,
    n_iter_no_change=15,
    random_state=RANDOM_SEED,
)

mlp_model.fit(X_train_nn_scaled, y_train_nn)

print(
    f"MLP trained: {mlp_model.hidden_layer_sizes} hidden layers, "
    f"{mlp_model.n_iter_} iterations, "
    f"final training loss: {mlp_model.loss_:.4f}"
)

# Predict full user x item matrix in batches for top-N ranking
pred_nn_matrix = np.zeros((n_users, n_items), dtype=np.float32)
all_items_arr = np.arange(n_items)
NN_BATCH_SIZE = 500

for start in range(0, n_users, NN_BATCH_SIZE):
    end = min(start + NN_BATCH_SIZE, n_users)
    batch_users = np.arange(start, end)

    u_grid = np.repeat(batch_users, n_items)
    i_grid = np.tile(all_items_arr, len(batch_users))

    X_batch = build_nn_features(u_grid, i_grid)
    X_batch_scaled = nn_scaler.transform(X_batch)

    preds = mlp_model.predict(X_batch_scaled)
    pred_nn_matrix[start:end] = preds.reshape(len(batch_users), n_items)

pred_nn_matrix = np.clip(pred_nn_matrix, 1.0, 5.0)

print("Neural network component ready")

# Predict Test Rating
print("\nGenerating MLP predictions for test data...")

mlp_test_predictions = []

for row in test_df.itertuples():
    pred_val = pred_nn_matrix[row.user_idx, row.item_idx]
    mlp_test_predictions.append(pred_val)

mlp_test_predictions = np.clip(
    np.array(mlp_test_predictions), 1.0, 5.0
)

y_true = test_df["rating"].values
y_pred = mlp_test_predictions

# Rating Prediction Metrics
rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mae = mean_absolute_error(y_true, y_pred)

# Classification (Rating >= 4.0 = Like, Rating < 4.0 = Not Like)
LIKE_THRESHOLD = 4.0

y_true_class = (y_true >= LIKE_THRESHOLD).astype(int)
y_pred_class = (y_pred >= LIKE_THRESHOLD).astype(int)

class_accuracy = accuracy_score(y_true_class, y_pred_class)
class_f1 = f1_score(y_true_class, y_pred_class, zero_division=0)

print("\n" + "-" * 60)
print("MLP Rating Prediction Result")
print("-" * 60)
print(f"RMSE           = {rmse:.4f}")
print(f"MAE            = {mae:.4f}")
print(f"Class Accuracy = {class_accuracy:.4f}")
print(f"Class F1-Score = {class_f1:.4f}")
print("\n")

# Confusion Matrix
cm = confusion_matrix(y_true_class, y_pred_class)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm, display_labels=["Not Like", "Like"]
)

fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(
    ax=ax, values_format="d", colorbar=True, cmap=plt.cm.Blues
)

ax.grid(False)
plt.title("MLP: Confusion Matrix", fontsize=12, pad=10)
plt.tight_layout()
plt.show()

print("\n")

# Actual vs Predicted Ratings Scatter Plot
plt.figure(figsize=(7, 6))
plt.scatter(y_true, y_pred, alpha=0.25, color="#1f77b4")
plt.plot([1, 5], [1, 5], linestyle="--", color="red")
plt.xlabel("Actual Rating")
plt.ylabel("Predicted Rating")
plt.title("MLP Neural Network: Actual vs Predicted Ratings")
plt.xlim(1, 5)
plt.ylim(1, 5)
plt.tight_layout()
plt.show()

# Top-N Evaluation For MLP Only
TOP_K = 5

print("\n" + "-" * 60)
print(f"MLP TOP-{TOP_K} RECOMMENDATION EVALUATION")
print("-" * 60)

# Items already seen in training
train_items_by_user = (
    train_df.groupby("user_idx")["item_idx"].apply(set).to_dict()
)

# Relevant test items (Rating >= 4.0)
test_relevant_by_user = (
    test_df[test_df["rating"] >= LIKE_THRESHOLD]
    .groupby("user_idx")["item_idx"]
    .apply(set)
    .to_dict()
)

# Evaluation containers
precision_scores = []
recall_scores = []
f1_scores = []
hit_scores = []
ndcg_scores = []
evaluated_users = 0

# Evaluate every user
for user_idx, relevant_items in test_relevant_by_user.items():

    if user_idx not in train_items_by_user:
        continue

    seen_items = train_items_by_user[user_idx]

    # Candidate items = items NOT seen during training
    candidate_items = [i for i in range(n_items) if i not in seen_items]

    if len(candidate_items) == 0:
        continue

    # Rank candidate items using MLP predicted rating
    candidate_scores = [
        (i, pred_nn_matrix[user_idx, i]) for i in candidate_items
    ]

    candidate_scores.sort(key=lambda x: x[1], reverse=True)

    recommended_items = [
        item for item, score in candidate_scores[:TOP_K]
    ]

    recommended_set = set(recommended_items)
    relevant_set = set(relevant_items)

    # Number of hits
    hits = len(recommended_set.intersection(relevant_set))

    # Precision@K
    precision = hits / TOP_K

    # Recall@K
    recall = hits / len(relevant_set) if len(relevant_set) > 0 else 0

    # F1@K
    if precision + recall > 0:
        f1 = 2 * precision * recall / (precision + recall)
    else:
        f1 = 0

    # Hit Rate@K
    hit = 1 if hits > 0 else 0

    # NDCG@K
    dcg = 0.0
    for rank, item in enumerate(recommended_items, start=1):
        if item in relevant_set:
            dcg += 1 / np.log2(rank + 1)

    ideal_hits = min(len(relevant_set), TOP_K)
    idcg = sum(1 / np.log2(rank + 1) for rank in range(1, ideal_hits + 1))

    ndcg = dcg / idcg if idcg > 0 else 0

    # Store results
    precision_scores.append(precision)
    recall_scores.append(recall)
    f1_scores.append(f1)
    hit_scores.append(hit)
    ndcg_scores.append(ndcg)

    evaluated_users += 1

# Display Top-N Results
mlp_precision = np.mean(precision_scores)
mlp_recall = np.mean(recall_scores)
mlp_f1 = np.mean(f1_scores)
mlp_hr = np.mean(hit_scores)
mlp_ndcg = np.mean(ndcg_scores)

print(f"Users evaluated = {evaluated_users}")
print(f"Precision@{TOP_K} = {mlp_precision:.4f}")
print(f"Recall@{TOP_K}    = {mlp_recall:.4f}")
print(f"F1@{TOP_K}        = {mlp_f1:.4f}")
print(f"HR@{TOP_K}        = {mlp_hr:.4f}")
print(f"NDCG@{TOP_K}      = {mlp_ndcg:.4f}")

# Sample Tourist Recommendation
print("\n" + "=" * 60)
print("Sample Tourist Recommendation")
print("=" * 60)

# Select a representative tourist
user_interaction_counts = train_df.groupby("user_idx").size()

# Select a Tourist with 8-12 Previous Interactions.
eligible_users = user_interaction_counts[
    (user_interaction_counts >= 8) & (user_interaction_counts <= 12)
]

# Randomly Select one Tourist
SAMPLE_USER_IDX = eligible_users.sample(
    1, random_state=RANDOM_SEED
).index[0]

# Convert user_idx Back to original tourist_id
tourist_id = user_ids[SAMPLE_USER_IDX]

print(f"\nTourist ID: {tourist_id}")
print(
    f"Previous interactions: {user_interaction_counts[SAMPLE_USER_IDX]}"
)

# Find previous attractions
previous_items = train_df[
    train_df["user_idx"] == SAMPLE_USER_IDX
].sort_values("rating", ascending=False)

print("\nPrevious Attractions")
print("-" * 70)
print(f"{'Attraction':<45}{'Rating':>10}")
print("-" * 70)

for row in previous_items.itertuples():
    attraction_name = idx_to_item[row.item_idx]
    print(f"{attraction_name:<45}{row.rating:>10.2f}")

# Get attractions already visited
seen_items = set(previous_items["item_idx"])

# Generate candidate attractions
candidate_items = [
    item_idx for item_idx in range(n_items) if item_idx not in seen_items
]

# Predict ratings for candidate attractions
recommendation_scores = []

for item_idx in candidate_items:
    predicted_rating = pred_nn_matrix[SAMPLE_USER_IDX, item_idx]
    recommendation_scores.append((item_idx, predicted_rating))

# Sort attractions by predicted rating
recommendation_scores.sort(key=lambda x: x[1], reverse=True)

# Select Top 5 recommendations
top_recommendations = recommendation_scores[:5]

# Display recommendations
print("\nRecommended Attractions")
print("-" * 70)
print(
    f"{'Rank':<8}{'Attraction':<45}{'Predicted Rating':>17}"
)
print("-" * 70)

for rank, (item_idx, predicted_rating) in enumerate(
    top_recommendations, start=1
):
    attraction_name = idx_to_item[item_idx]
    print(
        f"{rank:<8}{attraction_name:<45}{predicted_rating:>17.2f}"
    )

# Create a recommendation DataFrame
recommendation_results = []

for rank, (item_idx, predicted_rating) in enumerate(
    top_recommendations, start=1
):
    recommendation_results.append(
        {
            "Tourist ID": tourist_id,
            "Rank": rank,
            "Attraction": idx_to_item[item_idx],
            "Predicted Rating": round(predicted_rating, 2),
        }
    )

recommendation_results_df = pd.DataFrame(recommendation_results)

# Final MLP Model Summary
print("\n" + "=" * 60)
print("MLP FINAL SUMMARY")
print("=" * 60)
print(f"RMSE                    : {rmse:.4f}")
print(f"MAE                     : {mae:.4f}")
print(f"Classification Accuracy : {class_accuracy:.4f}")
print(f"Classification F1       : {class_f1:.4f}")
print(f"Precision@5             : {mlp_precision:.4f}")
print(f"Recall@5                : {mlp_recall:.4f}")
print(f"F1@5                    : {mlp_f1:.4f}")
print(f"HR@5                    : {mlp_hr:.4f}")
print(f"NDCG@5                  : {mlp_ndcg:.4f}")
print("=" * 60)
```

```python
# NORMALIZE COMPONENT SCORES
print("\n" + "=" * 60)
print("NORMALIZING COMPONENT SCORES")
print("=" * 60)

def minmax_normalize(matrix):
    min_vals = matrix.min(
        axis=1,
        keepdims=True
    )
    max_vals = matrix.max(
        axis=1,
        keepdims=True
    )
    denom = (
        max_vals - min_vals
    )
    denom[
        denom == 0
    ] = 1.0
    return (
        matrix - min_vals
    ) / denom

# Normalize CF scores
cf_norm = minmax_normalize(
    pred_cf_matrix
)

# Normalize content similarity
content_norm = minmax_normalize(
    content_sim_matrix
)

# Normalize the neural network scores
nn_norm = minmax_normalize(
    pred_nn_matrix
)

print(
    f"CF normalized range:      "
    f"{cf_norm.min():.4f} - "
    f"{cf_norm.max():.4f}"
)

print(
    f"Content normalized range: "
    f"{content_norm.min():.4f} - "
    f"{content_norm.max():.4f}"
)

print(
    f"NN normalized range:      "
    f"{nn_norm.min():.4f} - "
    f"{nn_norm.max():.4f}"
)
```

```python
# POPULARITY BASELINE
print("\n" + "=" * 60)
print("POPULARITY BASELINE")
print("=" * 60)

# Average rating for each attraction
item_mean_rating = (
    train_df
    .groupby("item_idx")["rating"]
    .mean()
)

# Number of ratings for each attraction
item_rating_count = (
    train_df
    .groupby("item_idx")["rating"]
    .count()
)

# Popularity score
popularity_score = (
    item_mean_rating
    * np.log1p(item_rating_count)
)

popularity_vector = np.full(
    n_items,
    popularity_score.mean()
)

for item_idx, score in (
    popularity_score.items()
):

    popularity_vector[item_idx] = score

# Normalize popularity score
popularity_norm_vector = (
    popularity_vector
    - popularity_vector.min()
)

max_popularity = (
    popularity_norm_vector.max()
)

if max_popularity != 0:

    popularity_norm_vector /= (
        max_popularity
    )

print("Popularity baseline ready")

```

```python
# TOP-N EVALUATION SETUP
print("\n" + "=" * 60)
print("TOP-N EVALUATION SETUP")
print("=" * 60)

# Items already seen during training
# These items should not be recommended again.
train_seen = train_df.groupby("user_idx")["item_idx"].apply(set).to_dict()

# Define relevant items
# An item is relevant if:
# Actual test rating >= 4.0
# 5 = Relevant
# 4 = Relevant
# 3 = Not Relevant
# 2 = Not Relevant
# 1 = Not Relevant
test_relevant = {}

for uid, group in test_df.groupby("user_idx"):
    relevant_items = group.loc[
        group["rating"] >= LIKE_THRESHOLD, "item_idx"
    ].tolist()

    test_relevant[uid] = set(relevant_items)

# Evaluation statistics
users_with_relevant_items = sum(
    len(items) > 0 for items in test_relevant.values()
)

total_relevant_items = sum(
    len(items) for items in test_relevant.values()
)

print(f"Like threshold:            {LIKE_THRESHOLD:.1f}")
print(f"Users with relevant items: {users_with_relevant_items:,}")
print(f"Total relevant test items: {total_relevant_items:,}")

# METRIC FUNCTIONS
def precision_at_k(recommended, relevant, k):
    recommended = recommended[:k]

    if len(recommended) == 0:
        return 0.0

    hits = len(set(recommended) & set(relevant))

    return hits / k

def recall_at_k(recommended, relevant, k):
    if len(relevant) == 0:
        return 0.0

    recommended = recommended[:k]

    hits = len(set(recommended) & set(relevant))

    return hits / len(relevant)

def f1_score_at_k(precision, recall):
    if (precision + recall) == 0:
        return 0.0

    return 2 * precision * recall / (precision + recall)

def hit_rate_at_k(recommended, relevant, k):
    recommended = recommended[:k]

    hits = len(set(recommended) & set(relevant))

    return 1.0 if hits > 0 else 0.0

def ndcg_at_k(recommended, relevant, k):
    recommended = recommended[:k]
    dcg = 0.0

    for rank, item in enumerate(recommended, start=1):
        if item in relevant:
            dcg += 1 / np.log2(rank + 1)

    ideal_hits = min(len(relevant), k)

    if ideal_hits == 0:
        return 0.0

    idcg = sum(1 / np.log2(rank + 1) for rank in range(1, ideal_hits + 1))

    return dcg / idcg

# EVALUATION ENGINE
def evaluate_ranking(score_matrix=None, score_vector=None):
    results = {
        k: {
            "Precision": [],
            "Recall": [],
            "F1": [],
            "HitRate": [],
            "NDCG": [],
        }
        for k in TOP_K_VALUES
    }

    # Evaluate each user
    for uid, relevant in test_relevant.items():
        # Skip users with no liked items in the test set
        if len(relevant) == 0:
            continue
        # Get scores
        if score_matrix is not None:
            user_scores = score_matrix[uid].copy()

        elif score_vector is not None:
            user_scores = score_vector.copy()

        else:
            raise ValueError(
                "Either score_matrix or score_vector must be provided."
            )

        # Remove training items-
        seen = train_seen.get(uid, set())

        if len(seen) > 0:
            user_scores[list(seen)] = -np.inf

        # Rank attractions
        ranked_items = np.argsort(user_scores)[::-1]

        # Calculate metrics
        for k in TOP_K_VALUES:
            precision = precision_at_k(ranked_items, relevant, k)
            recall = recall_at_k(ranked_items, relevant, k)
            f1 = f1_score_at_k(precision, recall)
            hit_rate = hit_rate_at_k(ranked_items, relevant, k)
            ndcg = ndcg_at_k(ranked_items, relevant, k)

            results[k]["Precision"].append(precision)
            results[k]["Recall"].append(recall)
            results[k]["F1"].append(f1)
            results[k]["HitRate"].append(hit_rate)
            results[k]["NDCG"].append(ndcg)

    # Average metrics across users
    summary = {}

    for k in TOP_K_VALUES:
        summary[k] = {}

        for metric, values in results[k].items():
            if len(values) > 0:
                summary[k][metric] = np.mean(values)
            else:
                summary[k][metric] = 0.0

    return summary
```

```python
# RATING PREDICTION & CLASSIFICATION METRICS

print("\n" + "=" * 60)
print("RATING PREDICTION & CLASSIFICATION METRICS")
print("=" * 60)

# Test data
test_users = test_df["user_idx"].values
test_items = test_df["item_idx"].values
y_true = test_df["rating"].values

# Binary ground truth
# Rating >= 4 = Liked
# Rating < 4  = Not liked
y_true_binary = (y_true >= LIKE_THRESHOLD).astype(int)

# Predictions
svd_preds = pred_cf_matrix[test_users, test_items]
content_preds = pred_content_matrix[test_users, test_items]
nn_preds = pred_nn_matrix[test_users, test_items]
hybrid_pred_matrix = (
    best_alpha * pred_cf_matrix
    + (1 - best_alpha) * pred_content_matrix
)

hybrid_preds = hybrid_pred_matrix[test_users, test_items]

# RATING MODEL EVALUATION FUNCTION
def evaluate_rating_model(name, continuous_preds):

    # RMSE
    rmse = np.sqrt(mean_squared_error(y_true, continuous_preds))

    # MAE
    mae = mean_absolute_error(y_true, continuous_preds)

    # Convert predicted rating to binary classification
    binary_preds = (continuous_preds >= LIKE_THRESHOLD).astype(int)

    # Accuracy
    correct = np.sum(y_true_binary == binary_preds)
    acc = correct / len(y_true_binary)

    # Confusion matrix components
    tp = np.sum((y_true_binary == 1) & (binary_preds == 1))
    fp = np.sum((y_true_binary == 0) & (binary_preds == 1))
    fn = np.sum((y_true_binary == 1) & (binary_preds == 0))

    # Precision
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0

    # Recall
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0

    # F1-score
    f1 = (
        2 * precision * recall / (precision + recall)
        if (precision + recall) > 0
        else 0
    )

    print(f"{name}:")
    print(f"    RMSE = {rmse:.4f} | MAE = {mae:.4f}")
    print(f"    Class Accuracy = {acc:.4f} | Class F1-Score = {f1:.4f}")
    print()


# Evaluate SVD
evaluate_rating_model("SVD Collaborative", svd_preds)

# Evaluate TF-IDF
evaluate_rating_model("TF-IDF Content", content_preds)

# Evaluate Neural Network
evaluate_rating_model("Neural Network (MLP + embeddings)", nn_preds)

# Evaluate Hybrid
evaluate_rating_model(f"Hybrid (alpha={best_alpha:.2f})", hybrid_preds)
```

```python
# Final Model Comparison
print("\n" + "=" * 60)
print("FINAL MODEL COMPARISON")
print("=" * 60)

# SVD
svd_metrics = evaluate_ranking(
    score_matrix=cf_norm
)

# TF-IDF Content-Based
content_metrics = evaluate_ranking(
    score_matrix=content_norm
)

# Neural Network
nn_metrics = evaluate_ranking(
    score_matrix=nn_norm
)

# Hybrid

hybrid_final_scores = (
    best_alpha * cf_norm + (1 - best_alpha) * content_norm
)

hybrid_metrics = evaluate_ranking(
    score_matrix=hybrid_final_scores
)

# Popularity

popularity_metrics = evaluate_ranking(
    score_vector=popularity_norm_vector
)

# Create comparison table
comparison_rows = []

for name, metrics in [
    (
        "Popularity",popularity_metrics
    ),
    (
        "SVD Collaborative Filtering",
        svd_metrics
    ),
    (
        "TF-IDF Content-Based",content_metrics
    ),
    (
        "Neural Network (MLP + embeddings)",nn_metrics
    ),
    (
        f"Hybrid (alpha={best_alpha:.2f})",hybrid_metrics
    )
]:
    comparison_rows.append({
        "Model": name,
        "Prec@5":
            metrics[5]["Precision"],
        "Rec@5":
            metrics[5]["Recall"],
        "F1@5":
            metrics[5]["F1"],
        "HR@5":
            metrics[5]["HitRate"],
        "NDCG@5":
            metrics[5]["NDCG"],
        "Prec@10":
            metrics[10]["Precision"],
        "Rec@10":
            metrics[10]["Recall"],
        "F1@10":
            metrics[10]["F1"],
        "HR@10":
            metrics[10]["HitRate"],
        "NDCG@10":
            metrics[10]["NDCG"]
    })
comparison_df = pd.DataFrame(
    comparison_rows
)
print(
    comparison_df.to_string(
        index=False,
        float_format=lambda x:
            f"{x:.4f}"
    )
)
```
